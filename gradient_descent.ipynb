```python
import numpy as np

f = lambda p: p[0]**2 + p[1]**2
p = np.array([1.0, 2.0])
eps = 0.01

# 只用符号:方向 (1, 1)
d_sign = np.array([1, 1]) / np.sqrt(2)
# 用梯度数值:方向 (2, 4)
d_grad = np.array([2, 4]) / np.linalg.norm([2, 4])

print(f(p + eps * d_sign) - f(p))  # ≈ 0.0424
print(f(p + eps * d_grad) - f(p))  # ≈ 0.0447(更大,最陡)
```

```python
# y = w·x 拟合,loss = (w·x − y)²
x = np.array([1., 2., 3., 4.])
y = np.array([2., 4., 6., 8.])   # 真实 w = 2
w = 0.0

# see all examples:4 个样本梯度取平均
grad_full = np.mean(2 * (w * x - y) * x)   # -30.0

# see only one example:只看第 1 个样本
grad_one = 2 * (w * x[0] - y[0]) * x[0]    # -4.0

print(grad_full, grad_one)
```
