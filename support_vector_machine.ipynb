## Core idea
Find an optimal Hyperplane(一条线/一个面) to divide categories.

## Terms
* **Support vectors** are the data points closest to the hyperplane —
  they are the ones that determine its position/orientation (other
  points don't affect the boundary)
* **Margin** is the distance between the hyperplane and the nearest
  support vector on each side (SVM maximizes this margin)

## Kernel functions
Compute the dot product data would have *if* mapped into a higher
dimension — without actually doing the mapping (the "kernel trick"),
saving computation.

### Polynomial kernel
$$K(a, b) = (a \cdot b + r)^d$$
The result (a scalar) is used in place of the dot product inside the
SVM optimization (solved via Lagrange multipliers).

### Radial basis function (RBF) kernel
$$K(a, b) = e^{-\gamma \|a-b\|^2}$$
Corresponds to an infinite-dimensional feature space; $\gamma$ controls
how far the influence of a single point reaches (large $\gamma$ →
tighter/more local boundary).
