# Ridge/Squared/L2 Regularization
$$
\min_{\vec{w},b} J(\vec{w}, b) = \min_{\vec{w},b} \left[ \underbrace{\frac{1}{2m}\sum_{i=1}^{m}\left(f_{\vec{w},b}(\vec{x}^{(i)}) - y^{(i)}\right)^2}_{\text{fit data}} + \underbrace{\frac{\lambda}{2m}\sum_{j=1}^{n} w_j^2}_{\text{keep } w_j \text{ small}} \right]
$$

- **1st term (MSE):** fit the data
- **2nd term (regularization):** keep $w_j$ small
- **λ:** weight of the regularization, balances both goals (large λ → underfitting, small λ → overfitting)

Example (degree-4 polynomial):
$$
f_{\vec{w},b}(\vec{x}) = w_1 x + w_2 x^2 + w_3 x^3 + w_4 x^4 + b
$$

When apply ridge regularization to models with multiple parameters, parameters more associative will shrink little while non-associative parameters will shrink a lot. Because shrinking non-associative parameters will decrease penealty and not affect SSR/MSE a lot so the total score will be minimized.

Lasso regularization is to replace the squared penalty with absolute penalty. Lasso regularization can minimize a param into 0.
