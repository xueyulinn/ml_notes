## Classfication Trees
### Why
If we fit a s-shaped squiggle with logistic regression, there would be lots of misclassfications.

A classfication tree can handle all types of data, all types of relationships among the independent variables.

## Classification trees
### Step-by-step
1. For each candidate split (feature/threshold), compute the resulting
   nodes' impurity (e.g. Gini impurity)
2. Pick the split that gives the **lowest weighted impurity** (i.e. the
   largest impurity reduction) — this becomes the root/current node's split
3. For numeric features, candidate thresholds are the averages of
   adjacent sorted values
4. Recurse on each child node until a stopping condition (max depth,
   min samples, pure node, etc.)

## Regression tree
### Step-by-step
1. For each candidate threshold, split data and calculate SSR (sum of
   squared residuals) on each side, using each side's mean as prediction
2. Pick the threshold that minimizes the **total SSR (sum of both sides)**
   as the current split
3. Recurse on each side until a stopping condition is met (e.g. number
   of samples in a node < minimum, max depth)
