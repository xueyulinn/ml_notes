## 贝叶斯公式
$$ 
\begin{align*}
P(B_{i}|A) &= \frac{P(A\cap B)}{P(A)}\\
&= \frac{P(A\cap B)}{P(A|B_{1})P(B_{1}) + P(A|B_{2})P(B_{2}) + ... + P(A|B_{n})P(B_{n})}
\end{align*}
$$
分布是全概率公式, $P(B_{i})$是先验prior, $P(B_{i}|A)$是后验posterior

| | PMF | PDF |
|---|---|---|
| 适用 | 离散随机变量 | 连续随机变量 |
| P(X=x)的值 | 就是概率(0~1之间) | f(x)不是概率，可以>1 |
| 归一化条件 | $\sum_x P(X=x)=1$ | $\int_{-\infty}^{\infty} f(x)\,dx=1$ |
| 求区间概率 | 直接求和 | 需要积分 |

## 独立和条件独立
A和B独立
$$ P(A\cap B) = P(A)P(B) $$
A和B关于C独立
$$ P(A\cap B \mid C) = P(A|C)P(B|C) $$
A和B独立并不一定关于C独立

假设抛两次硬币, A是第一次头朝上, B是第二次头朝上, C是两次结果不一致, P(A) > 0, P(B) > 0, $P(A\cap B | C) = 0 $, P(A\cap B | C) != P(A)P(B)

## 多元随机变量
### 边缘分布列
$$
\begin{align*} 
P_{X}(x) &= \sum_{y} P_{X,Y}(x,y) \\
&= \sum_{y} P_{X|Y}(x|y)P_{Y}(y)
\end{align*} 
$$
### 条件分布列
$$ P_{X,Y}(x,y) = P_{Y}(y) P_{X|Y}(x,y) $$
$$ P_{X,Y}(x,y) = P_{X}(x) P_{Y|X}(y, x) $$

### 独立随机变量的预期值
$$ E(XY) = E(X)E(Y) $$
$$ Var(X+Y) = Var(X) + Var(Y) $$

### 随机变量和事件独立
$$ P_{X}(x) = P_{X|A}(x) $$
### 随机变量之间的独立性
如果随机变量X和Y独立, 则
$$ P_{X}(x) = P_{X|Y}(x|y) $$
### 条件独立
$$ P(X=x Y=y) = P(X=x|A)P(Y=y|A) $$

### 协方差
$$ cov(X,Y) = E[(X−E[X])(Y−E[Y])] $$
### 多个随机变量
两两分析协方差, xixj, 使用**协方差矩阵**, xixi就是方差
### 相关系数
$$ p_{X,Y} = \frac{cov(X,Y)}{\sigma{X} \sigma{y}} $$
消除量纲对协方差的影响

## 大数定律
对于独立同分布的随机变量$ X_{1},X_{2},X_{3}...X_{n} $, $ M_{n} = \frac{X_{1},X_{2},X_{3}...X_{n}}{n} $
$$
E[M_n] = E\left[\frac{X_1+X_2+\cdots+X_n}{n}\right] = \frac{1}{n}\Big(E[X_1]+E[X_2]+\cdots+E[X_n]\Big) = \frac{1}{n}\times n \times \mu = \mu = E[X_i]
$$
$$
\begin{align*}
\text{var}[M_n] &= \text{var}\left[\frac{X_1+X_2+\cdots+X_n}{n}\right] \\
&= \frac{1}{n^2}\text{var}[X_1+X_2+\cdots+X_n]\\ 
&= \frac{1}{n^2}\Big(\text{var}[X_1]+\text{var}[X_2]+\cdots+\text{var}[X_n]\Big) \\
&= \frac{1}{n^2}\times n \times \sigma^2 = \frac{\sigma^2}{n}
\end{align*}
$$
所以当 $n \to \infty$, $\text{var}[M_n] \to 0$，当样本数量足够大，样本均值 $M_n$（这个随机变量本身）就越来越集中在 $\mu$ 附近。

## 中心极限定理

获取随机变量 $X_1, X_2, \dots, X_n$，这 $n$ 个随机变量满足独立同分布，均值为 $\mu$、方差为 $\sigma^2$。

我们在这组随机变量的基础之上得到一个新的随机变量：

$$Z_n = \frac{X_1+X_2+\cdots+X_n-n\mu}{\sqrt n\sigma}$$

$$E[Z_n] = E\left[\frac{X_1+X_2+\cdots+X_n-n\mu}{\sqrt n\sigma}\right] = \frac{E[X_1+X_2+\cdots+X_n]-n\mu}{\sqrt n\sigma} = 0$$

$$\text{var}[Z_n] = \text{var}\left[\frac{X_1+X_2+\cdots+X_n-n\mu}{\sqrt n\sigma}\right] = \frac{\text{var}[X_1]+\text{var}[X_2]+\cdots+\text{var}[X_n]}{n\sigma^2} = 1$$

**中心极限定理（CLT）**：当 $n \to \infty$，$Z_n \sim N(0,1)$

## 随机过程
两种随机过程, 一种是到达过程, 关注的是某种"到达"事件是否发生, 相邻两次到达的时间为独立的变量, "无记忆". 可以是离散的, 随机过程就是伯努利trial, 服从几何分布, 连续的, 随机过程就是泊松过程, 服从指数分布

另一种不是独立的, 未来的数据与历史数据有关联, **马尔可夫过程**就属于这一种, 这个过程未来的数据只依赖于当前的数据，而与过去的历史数据无关.

## 马尔可夫链

### 离散的马尔可夫链

**三要素:**
- 离散时间 $n$
- 状态空间 $S=\{1,2,\dots,m\}$
- 转移概率 $p_{ij}$

**转移概率矩阵:**
所有满足 $p_{ij}>0$ 的状态对 $(i,j)$,可以组织成一个 $m\times m$ 的二维矩阵,第 $i$ 行第 $j$ 列为 $p_{ij}$。

约束条件:每一行元素之和为1,即对所有 $i$:
$$\sum_{j=1}^m p_{ij}=1$$

**马尔可夫性(核心定义):**
由三要素描述的随机变量序列 $X_0,X_1,X_2,X_3,\dots$ 取值于状态空间 $S$,满足对任意时刻 $n$、任意状态 $i,j\in S$,以及任意历史状态序列 $i_0,i_1,\dots,i_{n-1}$,均有:

$$P(X_{n+1}=j\mid X_n=i,X_{n-1}=i_{n-1},\dots,X_0=i_0)=P(X_{n+1}=j\mid X_n=i)=p_{ij}$$

即:下一状态只依赖当前状态,与历史无关。


### 多步转移

通过 $m$ 步(其中 $m>1$)从状态 $i$ 转移到状态 $j$,对应的就是 **$m$ 步状态转移概率**:

$$p^m(i,j) = P(X_{n+m}=j \mid X_n=i)$$

**计算方式(以 $m=2$ 为例):**
中间需经过某个状态 $k$,把所有可能路径的概率相乘后再求和:

$$p^2(i,j) = \sum_{k} p_{ik}\cdot p_{kj}$$

这正是矩阵乘法的定义,因此:**$m$ 步转移矩阵 = 一步转移矩阵的 $m$ 次方**,即 $P^m$。

### 常返态（Recurrent States）

设状态 $i$ 的可达状态集合为 $A(i)$。若对 $A(i)$ 中每个状态 $j$，都有 $i \in A(j)$（即 $j$ 也能返回 $i$），则称状态 $i$ 为**常返态**。

通俗地说：从 $i$ 出发能到达的每个状态，都能反过来回到 $i$。

### 非常返态（Transient States）

不满足上述条件的状态，即存在某个可达状态 $j$，使得从 $j$ 无法返回 $i$。

### 常返类与非常返类

- **常返类**：只进不出（closed class）——一旦进入该类，永远无法离开。
- **非常返类**：最终会离开，进入某个常返类，不会被再次访问。

### 周期性（Periodicity）

判断常返类 $R$ 是否为非周期的（aperiodic）：

若存在某个特定的 $n \geq 1$ 和 $R$ 中某个特定的状态 $i$，使得从状态 $i$ 出发经过恰好 $n$ 步，可以到达 $R$ 中的**任意**状态，那么这个常返类就是**非周期的**。

（等价地：类的周期 = 从任一状态返回自身所需步数的最大公约数；若周期为 1，则非周期。）

### 可约
不可约的马尔可夫链是指从任意一个状态出发，经过充分长的时间之后，可以到达任意状态，否则就是可约的马尔可夫链。 

### 稳态
对于马尔可夫链中的每一个状态j， n步转移概率值与(n)会趋近于一个独立于初始状态i的极限值，我们通常把这个极限值记作巧。到达稳态后，经过下一个时间的状态转移，马尔可夫链的概率分布保持不变，这就叫平稳。 

在不可约、非周期、正常返的条件下，马尔可夫链拥有唯一稳态分布，且每个状态的概率都大于0。

## 隐马尔科夫模型
隐马尔可夫模型是一个时序模型，首先它由一个隐藏的马尔可夫链按照设定的状态转移概率，随机生成一个状态随机序列，但是这个随机序列是无法观测到的，然后再由每个状态按照观测概率(或称输出概率)生成各自对应的观测，由此构成可观测的观测随机序列。
