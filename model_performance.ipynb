## Confusion matrix
A table that compares predicted labels vs actual labels for a
classification model, broken down into TP / FP / TN / FN.

|                | Predicted: Positive | Predicted: Negative |
|----------------|:--------------------:|:--------------------:|
| **Actual: Positive** | TP | FN |
| **Actual: Negative** | FP | TN |

## Metrics
### Sensitivity/Recall/True positive rate
$$ \frac{TP}{TP+FN} $$
How well a model correctly classifies positives.
### Specificity
$$ \frac{TN}{FP+TN} $$
How well a model correctly classifies negatives.
### Precision
$$ \frac{TP}{TP+FP} $$
Evaluate the quality of classfied positives.
### False positive rate
$$ \frac{FP}{FP+TN} $$
The percentage of misclassified positives.

## Receiver operating characteristic curve
1. Get data points(TPR, FPR) by changing threshold for a confusion matrix.
2. Connect all data points

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve, auc

# 假设你已有真实标签 y_true 和模型输出的概率 y_scores
y_true = np.array([0, 0, 1, 1, 1, 0, 1, 0])
y_scores = np.array([0.1, 0.3, 0.6, 0.8, 0.4, 0.2, 0.7, 0.5])

fpr, tpr, thresholds = roc_curve(y_true, y_scores)
roc_auc = auc(fpr, tpr)

plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='navy', lw=1, linestyle='--')  # 随机猜测的基准线
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('ROC Curve')
plt.legend(loc='lower right')
plt.show()
```

## Area under curve (AUC)
The area under the **ROC curve** (TPR vs FPR across different
classification thresholds), used to compare different models with a
single value. The larger the AUC, the better the model performs
(AUC = 1 is perfect, AUC = 0.5 is random guessing).

Ideally TPR = 1 and FPR = 0 (top-left point) — a curve that hugs the
top-left corner has the largest area under it.
