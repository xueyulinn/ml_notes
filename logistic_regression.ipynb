# 逻辑回归：从线性模型到 Sigmoid

**推导复习讲义**

---

## 1. 一句话主线

线性模型只会吐出 $(-\infty,\infty)$ 的分数 $z$，而概率必须落在 $(0,1)$。整条逻辑回归就是为了**弥合这个范围矛盾**而搭出来的：先把 $p$ 拉伸到实数轴，令其等于 $z$，再反解出 $p$，得到 sigmoid。

$$
\underbrace{z=\mathbf{w}\cdot\mathbf{x}+b}_{(-\infty,\infty)}
\;\xrightarrow{\;\text{范围对不上}\;}\;
\underbrace{\log\frac{p}{1-p}=z}_{\text{把 }p\text{ 拉到 }(-\infty,\infty)}
\;\xrightarrow{\;\text{取指数解 }p\;}\;
\underbrace{p=\frac{1}{1+e^{-z}}}_{\text{sigmoid}}
$$

## 2. $z$ 是什么

$z$ 是**线性预测量**，与线性回归算的是同一个东西：

$$z=\mathbf{w}\cdot\mathbf{x}+b .$$

对单个样本它是一个标量，取值范围 $(-\infty,\infty)$。它**不是**概率，只是打分：$z$ 越大越倾向 $y=1$，$z=0$ 对应 $p=0.5$。因它恰等于对数几率，也被称作 *logit*。

## 3. 核心矛盾：范围不匹配

| 量 | 表达式 | 取值范围 |
|---|---|---|
| 概率 $p$ | — | $(0,1)$ |
| 线性模型 $z$ | $\mathbf{w}\cdot\mathbf{x}+b$ | $(-\infty,\infty)$ |

天真地令 $p=z$ 会崩：无界的线性函数会输出 $1.7$ 或 $-0.3$，不是合法概率。故**不能**直接把两者划等号。

## 4. 搭桥：把 $p$ 从 $(0,1)$ 拉到 $(-\infty,\infty)$

思路是反过来——不压缩 $z$，而是拉伸 $p$。分两步逐个拆掉边界：

| 步骤 | 变换 | 范围 |
|---|---|---|
| 概率 → 几率 | $\text{odds}=\dfrac{p}{1-p}$ | $(0,\infty)$，拆掉上界 $1$ |
| 几率 → 对数几率 | $\log\dfrac{p}{1-p}$ | $(-\infty,\infty)$，拆掉下界 $0$ |

对数几率的范围与 $z$ 一致，等号至此**合法**：

$$\boxed{\;\log\frac{p}{1-p}=z=\mathbf{w}\cdot\mathbf{x}+b\;}$$

端点对应：$p\to 0 \Rightarrow z\to-\infty$；$p\to 1 \Rightarrow z\to+\infty$；$p=0.5 \Rightarrow z=0$。

## 5. 为什么偏偏是 log-odds

任何把 $(0,1)$ 单调双射到 $(-\infty,\infty)$ 的函数都能当桥（如 *probit*，用标准正态逆 CDF）。选 log-odds 有两个实打实的理由：

- **可解释性**：得到 $\text{odds}=e^{z}$，系数 $e^{w_j}$ 直接是 odds ratio。
- **典范链接函数**：对数几率正是伯努利分布的自然参数（指数族理论），因此是最"自然"的选择，而非"恰好能用"。

## 6. 反解出 $p$：Sigmoid 的诞生

关键一步是**对两边取指数消掉 log**（$e$ 与 $\ln$ 互逆，$e$ 在此诞生）：

$$
\begin{aligned}
\log\frac{p}{1-p}=z
&\;\xRightarrow{\;e^{(\cdot)}\;}\;
\frac{p}{1-p}=e^{z}\\[4pt]
&\Rightarrow\; p=e^{z}(1-p)=e^{z}-e^{z}p\\
&\Rightarrow\; p(1+e^{z})=e^{z}\\
&\Rightarrow\; p=\frac{e^{z}}{1+e^{z}}=\frac{1}{1+e^{-z}}=g(z).
\end{aligned}
$$

**一般底数版本**（说明 $e$ 从何而来）：若 logit 用底 $b$，$\log_b\frac{p}{1-p}=z\Rightarrow p=\dfrac{1}{1+b^{-z}}$。sigmoid 底数里的数，就是 logit 里对数的底；逻辑回归用 $\ln$，故底为 $e$。

## 7. Sigmoid 与 Logit 互为反函数

$$
\text{logit}:(0,1)\to(-\infty,\infty)
\qquad
\text{sigmoid}=\text{logit}^{-1}:(-\infty,\infty)\to(0,1).
$$

反函数对调定义域与值域——这正是下一节值域被"锁死"的根本原因。

## 8. 值域 $(0,1)$：有意为之，且必然锁死

**有意为之：** 幻灯片标题"Want outputs between 0 and 1"是*先提出的需求*，后续构造都是倒着为满足它而设计的，绝非巧合。

**证明一（反函数）：** logit 的定义域是 $(0,1)$，反函数把定义域与值域对调，故 sigmoid 值域*必然*为 $(0,1)$。

**证明二（直接看公式）：** 对任意实数 $z$，

$$
e^{-z}>0 \;\Rightarrow\; 1+e^{-z}>1>0
\;\Rightarrow\; 0<\frac{1}{1+e^{-z}}<1 .
$$

分母恒正 $\Rightarrow g>0$；分母恒 $>1 \Rightarrow g<1$。结构上就被夹在 $0$ 与 $1$ 之间。

**易错点：是开区间 $(0,1)$，不是闭区间 $[0,1]$。** $z\to+\infty$ 时 $g\to 1^{-}$，$z\to-\infty$ 时 $g\to 0^{+}$，均取不到端点。因为端点对应对数几率 $\pm\infty$，而有限的 $z=\mathbf{w}\cdot\mathbf{x}+b$ 到不了 $\pm\infty$，故模型**永不输出绝对确定的 $0$ 或 $1$**。

## 9. 可解释性小结（odds ratio）

由 $\text{odds}=e^{z}=e^{\mathbf{w}\cdot\mathbf{x}+b}$，得

$$
e^{w_j}=x_j\text{ 每增加一个单位时，几率 (odds) 乘上的倍数（odds ratio）}.
$$
