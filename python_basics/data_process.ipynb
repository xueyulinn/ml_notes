```python
import numpy as np
```

```python
''' 
init ndarray
'''
arr1 = np.array([1,2,3,4])
print(arr1, type(arr1))
arr2 = np.zeros([2,2,3], dtype=np.float32)
print(arr2)
arr3 = np.arange(1, 10)
print(arr3)
```

```python
''' 
modify ndarray
'''
print(1 + arr3)
print(1 / arr3)
# calculates the exponential e**x
print(np.exp(1))
print(np.exp(arr3))
```

```python
'''  
reshape ndarray
'''
arr4 = np.array([1,2,3,4])
print(arr4)
print(arr4.shape)
arr5 = arr4.reshape(2, 2)
print(arr5.shape)
```
