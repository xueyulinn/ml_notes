```python
'''
整数类型
CPython 内部不用机器原生的定宽整数来存 int，而是用一个结构体，
把大整数拆成若干"位块"（digit，通常每块 30 位）存进一个变长数组里——本质是软件实现的大整数
代价：Python 的整数运算比硬件原生整数慢得多，而且大整数越大越慢
struct _longobject {
    PyObject_VAR_HEAD          // 对象头：引用计数、类型指针、以及一个 size 字段
    digit ob_digit[1];         // 变长数组，真正存放"位块"的地方
};
'''
a = 22278787879877666
print(a)
```

```python
'''
is比较对象, ==比较值
'''
a = 1111
b = 1111
# is检查变量指向的对象
print(a is b)

c = 256
d = 256
# 缓存[-5,256]
print(c is d)

name1 = "Lee"
name2 = "Lee"

# CPython 会把一部分字符串放进一个全局的驻留池里复用。不可靠
print(name1 is name2)
# 判断值相等
print(name1 == name2)

a = "hello world!"
b = "hello world!"
print(a is b)
print(a == b)

# None是一个单例对象
```

```python
# OOP
class Example:
  pass

Example.count = 1
example = Example()
example.count = 3
print(example.count)
print(example.__class__.count)
```

```python
class Example:
  # ctor
  # 成员函数第一个参数是self
  def __init__(self, count, name=None):
    self.count = 10
    # private
    self.__name = name

example = Example(10)
print(example.count)
# can not get member prefix with __
print(example.__name)
```
